```python
import numpy as np

```

```python
def f(x):
    return -x**2 + 5*x + 20

# ---------- given data ----------
x = np.array([-9.6, -6, -2.6, -1.1, 0.6, 2.3, 2.8, 8.3, 10.0])
v = np.zeros_like(x)
c1 = c2 = 1.0
r1, r2 = 0.213, 0.876        # fixed as in the question
n_iter = 20
tol_f, tol_x = 1e-6, 1e-4    # convergence tolerances

```

```python
def print_table(it, x, v):
    fx = f(x)
    print(f"\nIteration {it}")
    print(f"{'Particle':<10}{'Position x':>14}{'Velocity v':>14}{'Fitness f(x)':>16}")
    print("-" * 54)
    for i in range(len(x)):
        print(f"{i+1:<10}{x[i]:>14.4f}{v[i]:>14.4f}{fx[i]:>16.4f}")

# ---------- initialisation ----------
```

```python
pbest = x.copy()
pbest_f = f(x)
g = np.argmax(pbest_f)
gbest, gbest_f = pbest[g], pbest_f[g]

print_table(0, x, v)
print(f"gbest = {gbest:.4f}, f(gbest) = {gbest_f:.4f}")

converged_at = None

```

```python
# ---------- main loop ----------
for t in range(1, n_iter + 1):
    v = v + c1*r1*(pbest - x) + c2*r2*(gbest - x)
    x = x + v

    fx = f(x)
    better = fx > pbest_f              # maximization
    pbest[better] = x[better]
    pbest_f[better] = fx[better]

    prev_gbest_f = gbest_f
    g = np.argmax(pbest_f)
    gbest, gbest_f = pbest[g], pbest_f[g]

    print_table(t, x, v)
    print(f"gbest = {gbest:.4f}, f(gbest) = {gbest_f:.4f}")

    df = abs(gbest_f - prev_gbest_f)
    spread = np.max(np.abs(x - gbest))
    print(f"Convergence check: |df| = {df:.2e}, max|x - gbest| = {spread:.2e}")

    if converged_at is None and df < tol_f and spread < tol_x:
        converged_at = t

print("\n" + "=" * 54)
print(f"Final gbest x = {gbest:.6f}, f = {gbest_f:.6f}")
print("Analytical optimum: x = 2.5, f = 26.25")
if converged_at:
    print(f"Convergence criterion met at iteration {converged_at}")
else:
    print(f"Convergence criterion (df<{tol_f}, spread<{tol_x}) not met within {n_iter} iterations")
```

```python
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt



def f(x): return -x**2 + 5*x + 20

x = np.array([-9.6, -6, -2.6, -1.1, 0.6, 2.3, 2.8, 8.3, 10.0])
v = np.zeros_like(x)
c1 = c2 = 1.0; r1, r2 = 0.213, 0.876
pbest, pbest_f = x.copy(), f(x)
g = np.argmax(pbest_f); gbest, gbest_f = pbest[g], pbest_f[g]
gf = [gbest_f]

for t in range(1, 11):                      # up to convergence (iteration 10)
    v = v + c1*r1*(pbest - x) + c2*r2*(gbest - x)
    x = x + v
    fx = f(x); b = fx > pbest_f
    pbest[b], pbest_f[b] = x[b], fx[b]
    g = np.argmax(pbest_f); gbest, gbest_f = pbest[g], pbest_f[g]
    gf.append(gbest_f)

it = np.arange(11)
plt.figure(figsize=(8, 5))
plt.plot(it, gf, "o-", color="tab:blue", label="gbest fitness f(gbest)")
plt.axhline(26.25, color="r", ls="--", label="optimum f = 26.25")
plt.axvline(10, color="g", ls=":", label="converged (iteration 10)")
plt.annotate(f"x = {gbest:.4f}\nf = {gbest_f:.4f}", (10, gf[-1]),
             xytext=(6.5, 26.215), arrowprops=dict(arrowstyle="->"))
plt.xlabel("Iteration"); plt.ylabel("f(gbest)")
plt.title("PSO convergence of gbest fitness (iterations 0-10)")
plt.xticks(it); plt.ylim(26.20, 26.26); plt.grid(alpha=.3); plt.legend(loc="lower right")
plt.tight_layout()

```
